```python
pd.read_sql_query("select count(*)from fraud_transactions",conn)
```

```python
pd.read_sql_query("""
select type, count(*) as totaltransactions
from fraud_transactions
group by type
""",conn)
```

```python
pd.read_sql_query("""
SELECT isFraud, COUNT(*) as count
FROM fraud_transactions
GROUP BY isFraud
""", conn)
```

```python
pd.read_sql_query("""
SELECT SUM(amount) as total_fraud_amount
FROM fraud_transactions
WHERE isFraud = 1
""", conn)
```

```python
pd.read_sql_query("""
SELECT type, COUNT(*) AS fraud_count
FROM fraud_transactions
WHERE isFraud = 1
GROUP BY type
ORDER BY fraud_count DESC
""", conn)
```

```python
pd.read_sql_query("""
SELECT nameOrig, nameDest, amount
FROM fraud_transactions
WHERE isFraud = 1
ORDER BY amount DESC
LIMIT 10
""", conn)
```

```python
pd.read_sql_query("""
SELECT AVG(amount) AS avg_fraud_amount
FROM fraud_transactions
WHERE isFraud = 1
""", conn)
```

```python
pd.read_sql_query("""
SELECT 
ROUND(
SUM(CASE WHEN isFraud=1 THEN 1 ELSE 0 END)*100.0/COUNT(*),2
) AS fraud_percentage
FROM fraud_transactions
""", conn)
```

```python
pd.read_sql_query("""
SELECT type, COUNT(*) AS fraud_count
FROM fraud_transactions
WHERE isFraud = 1
GROUP BY type
ORDER BY fraud_count DESC
""", conn)
```

```python
pd.read_sql_query("""
SELECT nameOrig, COUNT(*) AS fraud_count
FROM fraud_transactions
WHERE isFraud = 1
GROUP BY nameOrig
ORDER BY fraud_count DESC
LIMIT 10
""", conn)
```
